Variables in Python

    Variables in Python are used to store data values. They are created when you assign a value to them, and you don’t need to declare their type (Python is dynamically typed).


```python
place= 123
print(place)
```

Variable Naming Rules:
- Variable names can contain letters (a-z, A-Z), numbers (0-9), and underscores (_).
- Variable names must start with a letter or an underscore.
- Variable names are case-sensitive (Name and name are different).
- The name of the variable must not be any of Python's reserved words (keywords)

Keywords:

['False', 'None', 'True', 'and', 'as', 'assert', 'break', 'class', 'continue', 'def', 'del', 'elif', 'else', 'except', 'finally', 'for', 'from', 'global', 'if', 'import', 'in', 'is', 'lambda', 'nonlocal', 'not', 'or', 'pass', 'raise', 'return', 'try', 'while', 'with', 'yield']

Data Types in Python

Python has various built-in data types. Some common ones are:

- int: For integers (e.g., 1, -3, 100)
- float: For floating-point numbers (e.g., 3.14, -0.001)
- str: For strings (e.g., "Hello", "Python")
- bool: For boolean values (True or False)
-Complex numbers are expressed as (real part) + (imaginary part) j, where j denotes the imaginary unit. (e.g 10+3j)

```python
# int
n1= 123
print(type(n1))

# float 
n2= 123.90
print(type(n2))

# complex 
n3= 10+ 3j
print(type(n3))

# boolen  True and False
is_true= True
print(type(is_true))
```

```python
# string
student_name= 'raj'
print(type(student_name))
```

Tuple

    A tuple is a collection of items that is ordered and immutable (unchangeable). Tuples are similar to lists, but once a tuple is created, you cannot modify it. They are often used to group related data together.

    Syntax:
    my_tuple = (element1, element2, element3, ...)

```python
student_names= ('raj',"isha","ramu")
print(type(student_names))
```

```python
student_names= (",",)
print(type(student_names))
```

List

    A list is a collection of items that are ordered, mutable (changeable), and allow duplicate elements. Lists can hold items of different data types, such as integers, strings, or even other lists.

    Syntax: my_list = [element1, element2, element3, ...]

```python
student_names= ['raj',"isha","ramu",123,True, ("tina","bina"),[10,20,30]]
print(type(student_names))
```

Set

    A set is a collection of unique items that is unordered and unindexed. Sets do not allow duplicate values. Sets are useful for performing operations like union, intersection, and difference.
    Syntax:
    my_set = {element1, element2, element3, ...}

```python
student_names ={"ramu",10,20,10,100,"ramu"}
print(type(student_names))
print(student_names)
```

Dictionaries

    A dictionary in Python is a collection of key-value pairs. Each key in a dictionary is associated with a value, and you can retrieve or manipulate data using the key. Unlike lists and tuples, dictionaries are unordered and mutable (changeable).

    Syntax:
        my_dict = {
            "key1": "value1",
            "key2": "value2",
            "key3": "value3"
        }


```python
student_names={"grade":6,"name":"usha", "name":"raju"}  # {key:valu}
print(type(student_names))
print(student_names)
```

Mutable and Immuatble Datatypes:

1. Mutable
    - Mutable means the object can be changed after it is created. You can add, remove, or modify its elements without creating a new object. Example: list,set,dictionary(dict)

2. Immutable
    - Immutable means the object cannot be changed after it is created. If you try to modify it, Python creates a new object instead. Example : integer,float,bool,str,tuple,

```python
list1= [10,20,30,40]
print(id(list1))
```

```python
list1[0]=100
print(id(list1))
print(list1)
```

```python
dict1={"grade":9}
print(id(dict1))
print(dict1)

# change

dict1["grade"]=1
print(id(dict1))
print(dict1)
```

```python
numbers= {10,20,30,40,50}
print(numbers)
print(id(numbers))
numbers.add(500)
print(id(numbers))
print(numbers)
```
